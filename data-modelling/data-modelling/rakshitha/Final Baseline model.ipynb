```python
import pandas as pd
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.ensemble import RandomForestRegressor
```

```python
df = pd.read_excel('/content/edufeed_clean_data.xlsx')
df.dropna(inplace=True)
```

```python
stop_words = set(stopwords.words('english'))
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)
df['processed'] = df['comments'].apply(clean_text)
```

```python
tfidf = TfidfVectorizer(max_features=2000)
X = tfidf.fit_transform(df['processed'])
```

```python
y_sentiment = df['sentiment_label']
sent_model = LogisticRegression(max_iter=200)
sent_model.fit(X, y_sentiment)
```

```python
y_rating = df['star_rating']
rating_model = RandomForestRegressor(n_estimators=20, max_depth=10, random_state=42)
rating_model.fit(X, y_rating)
```

```python
def predict_feedback(text):
    text_clean = clean_text(text)
    text_vec = tfidf.transform([text_clean])
    sentiment = sent_model.predict(text_vec)[0]
    rating = rating_model.predict(text_vec)[0]
    return sentiment, round(rating, 2)
```

```python
predict_feedback("The faculty is very helpful but exams are tough")
```
