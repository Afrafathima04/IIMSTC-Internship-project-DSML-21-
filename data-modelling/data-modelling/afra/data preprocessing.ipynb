# Data Preprocessing Code  : 

# STEP 1. Import Libraries


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# STEP 2. Load Dataset


```python
df = pd.read_excel("RateMyProfessor_Raw_Dataset.xlsx")

print("Dataset Loaded Successfully")
print(df.shape)
print(df.head())
print(df.info())
```

# STEP 3. Basic Cleaning: Duplicates and Indexing


```python
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)
print(f"After removing duplicates: {df.shape}")
```

# STEP 4: Date Parsing and Feature Extraction


```python
df['post_date'] = pd.to_datetime(df['post_date'], format='mixed', errors='coerce')
df['post_year'] = df['post_date'].dt.year.astype('Int64')
df['post_month'] = df['post_date'].dt.month.astype('Int64')
df['date_missing_flag'] = df['post_date'].isna().astype(int)
```

```python

print(f"Date nulls after parsing: {df['post_date'].isna().sum()}")
```

# STEP 5: Handling Missing Values (Imputation)

5.1: take_again - fill with median

```python
df['take_again'].fillna(df['take_again'].median(), inplace=True)
print(f"✓ take_again: filled with median")
```

5.2: tag_professor - fill with 'No Tag'

```python
df['tag_professor'].fillna('No Tag', inplace=True)
print(f"✓ tag_professor: filled with 'No Tag'")
```


 5.3: categorical columns - fill with 'Not Reported'

```python
for col in ['attence', 'for_credits', 'would_take_agains', 'grades']:
    df[col].fillna('Not Reported', inplace=True)
print(f"✓ attence, for_credits, would_take_agains, grades: filled with 'Not Reported'")
```

 5.4: student ratings - fill with median

```python
df['student_star'].fillna(df['student_star'].median(), inplace=True)
df['student_difficult'].fillna(df['student_difficult'].median(), inplace=True)
print(f"✓ student_star, student_difficult: filled with median")
```

5.5: comments - fill with empty string

```python
df['comments'].fillna('', inplace=True)
print(f"✓ comments: filled with empty string")
```

5.6: word_comment - fill with 0

```python
df['word_comment'].fillna(0, inplace=True)
print(f"✓ word_comment: filled with 0")
```

5.7: name_not_onlines - fill with 'Unknown'

```python
df['name_not_onlines'].fillna('Unknown', inplace=True)
print(f"✓ name_not_onlines: filled with 'Unknown'")
```

```python
print(f"\nRemaining nulls (critical cols):")
print(df[['star_rating', 'student_star', 'comments', 'gender', 'race']].isnull().sum())
 
```

# STEP 6: CLEAN TEXT COLUMNS

```python
text_cols = ['professor_name', 'school_name', 'department_name', 'local_name', 'state_name']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
print(f"✓ Cleaned and title-cased: {', '.join(text_cols)}")
```

# STEP 7: GRADE GROUPING

```python
def clean_grade(g):
    if g in ['A+', 'A', 'A-']:       return 'A Range'
    if g in ['B+', 'B', 'B-']:       return 'B Range'
    if g in ['C+', 'C', 'C-']:       return 'C Range'
    if g in ['D', 'F']:              return 'D/F'
    if g in ['WD', 'INC', 'Not']:    return 'Other/WD'
    return g
```

```python
df['grade_group'] = df['grades'].apply(clean_grade)
print(f"✓ Grades grouped: {df['grade_group'].value_counts().to_dict()}")
 
```

# STEP 8: OUTLIER CAPPING (IQR METHOD)

```python
def iqr_cap(series):
    """Cap outliers using IQR method"""
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return series.clip(lower, upper)
```

```python
outlier_cols = ['star_rating', 'student_star', 'student_difficult', 'diff_index', 'num_student', 'word_comment']
 
```

```python
for col in outlier_cols:
    df[col] = iqr_cap(df[col])
print(f"✓ Outlier capping done for: {', '.join(outlier_cols)}")
```

# STEP 9: CREATE SENTIMENT LABEL

```python
def get_sentiment(star):
    if star >= 4.0:  return 'Positive'
    if star >= 3.0:  return 'Neutral'
    return 'Negative'
```

```python
df['sentiment_label'] = df['student_star'].apply(get_sentiment)
print(f"✓ Sentiment distribution:")
print(df['sentiment_label'].value_counts())
```

# STEP 10: CREATE HELP RATIO

```python

df['help_ratio'] = (
    df['help_useful'] /
    (df['help_useful'] + df['help_not_useful'] + 1e-6)
)
print(f"✓ Help ratio created (mean: {df['help_ratio'].mean():.3f})")
```

# STEP 111: CREATE DIFFICULTY CATEGORY

```python
def get_difficulty(d):
    if d <= 2.5:  return 'Easy'
    if d <= 3.5:  return 'Moderate'
    return 'Difficult'
 
df['difficulty_category'] = df['student_difficult'].apply(get_difficulty)
print(f"✓ Difficulty categories:")
print(df['difficulty_category'].value_counts())
```

# STEP 12: CREATE EXPERIENCE BUCKET

```python
def get_experience(years):
    if years <= 2:   return '0-2 yrs'
    if years <= 5:   return '3-5 yrs'
    if years <= 10:  return '6-10 yrs'
    return '10+ yrs'
 
df['experience_bucket'] = df['year_since_first_review'].apply(get_experience)
print(f"✓ Experience buckets created:")
print(df['experience_bucket'].value_counts())
```

# STEP 13: CREATE COMMENT LENGTH CATEGORY

```python
def get_comment_len(words):
    if words == 0:   return 'No Comment'
    if words <= 15:  return 'Short'
    if words <= 50:  return 'Medium'
    return 'Long'
```

```python
df['comment_length_cat'] = df['word_comment'].apply(get_comment_len)
print(f"✓ Comment length categories:")
print(df['comment_length_cat'].value_counts())
```

# STEP 14: CREATE TAG COUNT

```python
df['tag_count'] = df['tag_professor'].apply(
    lambda x: 0 if x == 'No Tag' else len(str(x).split('  '))
)
print(f"✓ Tag count created (mean: {df['tag_count'].mean():.2f})")
```

# STEP 15: CREATE COURSE MODE

```python
print("\n[STEP 14] CREATING COURSE MODE")
print("-" * 80)
 
df['course_mode'] = df['IsCourseOnline'].map({1: 'Online', 0: 'Offline'})
print(f"✓ Course mode distribution:")
print(df['course_mode'].value_counts())
 
```

# STEP 16: CONVERT TO CATEGORICAL

```python
cat_cols = ['sentiment_label', 'difficulty_category', 'grade_group',
            'experience_bucket', 'comment_length_cat', 'course_mode',
            'gender', 'race', 'attence', 'for_credits', 'would_take_agains']
```

```python
for col in cat_cols:
    df[col] = df[col].astype('category')
print(f"✓ Converted {len(cat_cols)} columns to categorical dtype")
```

# STEP 17: CONVERT BINARY TAGS TO INT

```python
binary_tag_cols = ['gives_good_feedback', 'caring', 'respected', 'participation_matters','clear_grading_criteria', 'skip_class', 'amazing_lectures', 'inspirational','tough_grader', 'hilarious', 'get_ready_to_read', 'lots_of_homework','accessible_outside_class', 'lecture_heavy', 'extra_credit','graded_by_few_things', 'group_projects', 'test_heavy',
'so_many_papers', 'beware_of_pop_quizzes', 'IsCourseOnline']
 
```

```python
for col in binary_tag_cols:
    df[col] = df[col].astype(int)
print(f"✓ Converted {len(binary_tag_cols)} binary tag columns to int")
```

# STEP 18: FINAL SUMMARY & VALIDATION

```python
print(f"\nDataset shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"\nCritical nulls left:")
print(df[['star_rating', 'student_star', 'gender', 'race', 'sentiment_label']].isnull().sum())
```

# STEP 19: OUTLIER STATISTICS

```python
numeric_cols = ['star_rating', 'student_star', 'student_difficult','diff_index', 'num_student', 'word_comment']
```

```python
print(f"\n{'Column':<25} {'Q1':>6} {'Q3':>6} {'Lower':>8} {'Upper':>8} {'Outliers':>10}")
print("-" * 65)
```

```python
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lo = Q1 - 1.5 * IQR
    hi = Q3 + 1.5 * IQR
    n = ((df[col] < lo) | (df[col] > hi)).sum()
    print(f"{col:<25} {Q1:>6.2f} {Q3:>6.2f} {lo:>8.2f} {hi:>8.2f} {n:>10}")
```
