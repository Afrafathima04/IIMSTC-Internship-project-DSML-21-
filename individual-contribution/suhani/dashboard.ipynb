# Student Feedback Analysis Dashboard
This notebook defines a Streamlit dashboard for analyzing student feedback. It includes metrics, sentiment distribution, and rating trend visualizations.


## Overview
- `dashboard_data`: static JSON-based metrics and sentiment values
- `Streamlit`: used for layout, cards, and charts
- `Run this app` with `streamlit run dashboard.py` or convert it to a Streamlit script for execution.


```python
import streamlit as st
```

## Data Definition
The dashboard uses a static JSON-style dictionary to store metrics and sentiment scores.


```python
dashboard_data = {
    "metrics": {
        "Total Responses": "1,284",
        "Average Rating": "4.2/5",
        "Response Rate": "78%",
        "Students Reached": "1648"
    },
    "sentiment": {
        "Positive": 60,
        "Negative": 25,
        "Neutral": 15
    }
}
```

## Page Setup
Configure the Streamlit page title and layout before rendering the dashboard.


```python
st.set_page_config(page_title="Feedback Dashboard", layout="wide")
```

## KPI Summary
Display the key performance indicators at the top of the dashboard.


```python
st.title("STUDENT FEEDBACK ANALYSIS DASHBOARD")
st.markdown("Overview of feedback collected across all courses this semester")
st.write("---")

col1, col2, col3, col4 = st.columns(4)
metrics = list(dashboard_data["metrics"].items())

col1.metric(metrics[0][0], metrics[0][1])
col2.metric(metrics[1][0], metrics[1][1])
col3.metric(metrics[2][0], metrics[2][1])
col4.metric(metrics[3][0], metrics[3][1])
st.write("---")
```

## Sentiment Distribution
Visualize the proportions of positive, negative, and neutral feedback using labeled columns.


```python
sent_col1, sent_col2, sent_col3 = st.columns(3)

with sent_col1:
    st.success(f"Positive Feedback: {dashboard_data['sentiment']['Positive']}%")
with sent_col2:
    st.error(f"Negative Feedback: {dashboard_data['sentiment']['Negative']}%")
with sent_col3:
    st.info(f"Neutral Feedback: {dashboard_data['sentiment']['Neutral']}%")
```

## Charts
Include a trend line and rating distribution chart to complete the dashboard.


```python
st.write("---")
left_chart, right_chart = st.columns(2)

with left_chart:
    st.subheader("Feedback Trend Over Time")
    st.line_chart([20, 30, 25, 45, 58, 25])

with right_chart:
    st.subheader("Rating Distribution")
    st.bar_chart([8, 14, 22, 35, 57])
```
